```python
# 13.	Write a function grade(score) that returns:
# a.	90 and above: "Distinction"
# b.	75 to 89: "First Class"
# c.	60 to 74: "Second Class"
# d.	below 60: "Fail"

def grade(scores):
  if scores>=90:
    return("Distinction")
  elif scores>=75:
    return("first class")
  elif scores>=89:
    return('first class')
  elif scores>=60:
    return("second class")
  elif scores>=74:
    return("second class")
  else:
    return('fail')
res=scores.apply(grade)
print(res)

```

```python
# 14.	Display how many students fall in each grade, then the same as percentages.
print("students fall in each grade",res.value_counts())
print()
print("percentage",res.value_counts(normalize=True))
```

```python
# 15.	Display the unique grades and the number of unique grades.
print("unique grade",res.unique())
print(" number of unique grades",res.nunique())
```

```python
# 16.	Create a new Series bonus_scores by adding 5 grace marks to every student, capped at 100. Confirm the original scores is unchanged
bonus_scores=lambda scores:scores+5
print(scores.apply(bonus_scores))
print()
print("original scores \n",scores)
```
